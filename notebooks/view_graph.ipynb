# See your graph in VS Code

Google's own Jupyter extension draws BigQuery and Spanner graph results as an interactive picture, the same one the BigQuery console shows. VS Code runs Jupyter notebooks, so no separate VS Code extension is needed.

**One-time setup**
1. In VS Code, install the **Jupyter** and **Python** extensions.
2. In a terminal: `pip install "bigquery-magics[spanner-graph-notebook]" jupyter` (or `pip install "graphkit[notebook]"`).
3. Sign in: `gcloud auth application-default login`.
4. Open this file and pick your Python environment as the kernel (top right).

**The rule for drawing:** the query must return whole paths as JSON, with `RETURN TO_JSON(p)`. Plain columns show as a table instead.

```python
%load_ext bigquery_magics
```

## BigQuery

Change `YOUR_PROJECT` below. Click a node to see its properties. You can also expand a node to pull in its neighbours.

```python
%%bigquery --graph --project YOUR_PROJECT
GRAPH `YOUR_PROJECT.vehicle_supply_chain.SupplyChain`
MATCH p = (s:Supplier)-[:SUPPLIES]->(:Part)-[:USED_IN]->(:VehicleModel)
WHERE s.name = 'Northstar Brakes'
RETURN TO_JSON(p) AS path
LIMIT 50
```

Recalls, the parts they name, and who made those parts:

```python
%%bigquery --graph --project YOUR_PROJECT
GRAPH `YOUR_PROJECT.vehicle_supply_chain.SupplyChain`
MATCH p = (r:Recall)-[:RECALLS_PART]->(:Part)-[:MADE_BY]->(:Supplier)
WHERE r.severity = 'HIGH'
RETURN TO_JSON(p) AS path
```

## Spanner

The same package has a Spanner magic. Fill in your instance and database.

```python
%load_ext spanner_graphs
```

```python
%%spanner_graph --project YOUR_PROJECT --instance YOUR_INSTANCE --database YOUR_DATABASE
GRAPH SupplyChain
MATCH p = (:Supplier)-[:SUPPLIES]->(:Part)
RETURN SAFE_TO_JSON(p) AS path_json
LIMIT 50
```
